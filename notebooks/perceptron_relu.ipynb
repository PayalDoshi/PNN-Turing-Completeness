```python
# Importing pytorch libraries 
import torch
import torch.nn as nn

# Create one perceptron
# Here we are defining our own neural network model called perceptron
class Perceptron(nn.Module):
    def __init__(self):     # Initialize the perceptron
        super().__init__()  # Initialize the normal pytorch neural-network functionality

        self.linear = nn.Linear(2, 1)
        self.relu = nn.ReLU()
        
# Defines what happens to the input
    def forward(self, x):
        z = self.linear(x)
        output = self.relu(z)
        return output
# Creates actual model
model = Perceptron()

# Give perceptron an input
x = torch.tensor([[1.0, 0.0]])

# Send this input through the model
output = model(x)

print(output)
```

```python
import torch
import torch.nn as nn

class Perceptron(nn.Module):
    def __init__(self):
        super().__init__()
        self.linear = nn.Linear(2, 1)
        self.relu = nn.ReLU()

    def forward(self, x):
        z = self.linear(x)
        output = self.relu(z)

        print("Before ReLU:", z)
        print("After ReLU:", output)

        return output

model = Perceptron()

x = torch.tensor([[1.0, 0.0]])

output = model(x)
```

```python
#Difference between numpy array and tensor
import numpy as np
import torch

# NumPy array
a = np.array([1, 2, 3])

# PyTorch tensor
t = torch.tensor([1, 2, 3])

print(a)
print(t)
```

##### A tensor is like a NumPy array designed for machine learning.
Array = container for numerical data.
Tensor = similar container, but with extra capabilities needed for deep learning, especially GPU computation and automatic gradients.

| Feature                                  | NumPy Array                 | PyTorch Tensor |
| ---------------------------------------- | --------------------------- | -------------- |
| Stores numbers                           | ✅                           | ✅              |
| Multiple dimensions                      | ✅                           | ✅              |
| Matrix operations                        | ✅                           | ✅              |
| Designed for general numerical computing | ✅                           | Can do it      |
| Designed for neural networks             | Not primarily               | ✅              |
| GPU computation                          | Limited/not native in NumPy | ✅              |
| Automatic gradients                      | ❌                           | ✅              |
| Used for neural-network training         | Not directly                | ✅              |
