```python
import torch
import torch.nn as nn

class Perceptron(nn.Module):
    def __init__(self):
        super().__init__()
        self.linear = nn.Linear(2, 1)

    def forward(self, x):
        return self.linear(x)

model = Perceptron()

x = torch.tensor([[1.0, 0.0]])

output = model(x)

print(output)
```

```python
print("Weights:", model.linear.weight)
print("Bias:", model.linear.bias)
```
